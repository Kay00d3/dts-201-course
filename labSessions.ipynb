```python
import statistics #imported the statistics module to use the median function
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

# get the number of people in the survey
count = len(ages)
print("The number of people in the survey is: ", count)

# get the total of the ages in the survey
total = sum(ages)
print("The total of the ages is: ", total)

# get the mean of the ages in the survey
mean = total/count

print("The mean age of the survey is: ", f"{mean:.2f}")
print("The youngest person in the survey is: ", min(ages))
print("The oldest person in the survey is: ", max(ages))

# get the median of the ages in the survey
median = statistics.median(ages)
print("The median age of the survey is: ", f"{median:.2f}")
```

```python
import pandas as pd

# Load dataset
df = pd.read_excel("50_startups_messy.xlsx")

# Report shape, missing counts, and duplicate count
print("Shape (Rows, Columns):", df.shape)
print("Missing Values per Column:")
print(df.isnull().sum())
print("Total Duplicate Rows:", df.duplicated().sum())

#dropped duplicates
df = df.drop_duplicates()

#Standardize 'State' column text
df['State'] = df['State'].astype(str).str.strip().str.title()

#Coerced Marketing Spend to numeric, converting non-numeric values to NaN
df['Marketing Spend'] = pd.to_numeric(df['Marketing Spend'], errors='coerce')

#Draw a simple random sample of 20 with random_state=7, and a stratified 20% sample by State.
srs = df.sample(n=20, random_state=7)
print(srs)

strat = df.groupby("State", group_keys=False ) .apply(lambda g: g.sample(frac=0.2))
print(strat)

#Create two simple random samples of 15 rows each, using different random states
s1 = df.sample(n=15, random_state=42)
s2 = df.sample(n=15, random_state=101)

#Create a dictionary comparing the means across population and both samples
compare = pd.DataFrame({
    "Population": [df["R&D Spend"].mean(), df["Marketing Spend"].mean()],
    "Sample 1":   [s1["R&D Spend"].mean(), s1["Marketing Spend"].mean()],
    "Sample 2":   [s2["R&D Spend"].mean(), s2["Marketing Spend"].mean()]
}, index=["R&D Spend", "Marketing Spend"])

# Print rounded to 2 decimal places
print(compare.round(2))
```
